7.Idetifying the number of missing values in each column

```python
print("the number of missing values in each column are:\n",df.isnull().sum())
```

8.Fill missing values in numerical columns using the mean.

```python
df.fillna(df.mean(numeric_only=True),inplace=True)
```

9.Fill missing values in categorical columns using the mode.


```python
cat_cols=df.select_dtypes(include='object').columns
df[cat_cols] = df[cat_cols].apply(lambda x:x.fillna(x.mode()[0]))
```

10.Verify that there are no missing values remaining.

```python
df.isnull().sum()
```

16.Create a histogram for the purchase amount column.

```python
plt.hist(df['purchase_amount'])
plt.xlabel('purchase_amount')
plt.ylabel('Frequency')
plt.title('Histogram of purchase_amount')
plt.show()
```

17.Create a bar chart showing the count of customers by product category.

```python
counts = df['region'].value_counts()
plt.bar(counts.index, counts.values)
plt.xlabel('Region')
plt.ylabel('Number of Customers')
plt.title('Customers by Region')
plt.xticks(rotation=45)
plt.show()
```

18.Create a box plot for the purchase amount column.

```python
plt.boxplot(df['purchase_amount'],vert=False)
plt.xlabel('purchase amount')
plt.title('Box plot for purchase amount')
plt.grid(True)
plt.show()
```

19.Create a scatter plot between age and purchase amount.(using matplotlib)

```python
plt.scatter(df['age'],df['purchase_amount'])
plt.xlabel('age')
plt.ylabel('purchase amount')
plt.title('Scatter plot for the age and purhcase amount')
plt.grid(True)
plt.show()
```

20.Display a correlation heatmap for numerical columns.(using sns)

```python
num_df = df[['age', 'annual_income', 'purchase_amount','loyalty_score', 'purchase_frequency']]
corr=num_df.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr,annot=True,cmap='plasma')
plt.title('Correlation Heatmap')
plt.show()
```
